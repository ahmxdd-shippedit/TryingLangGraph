```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
```

```python
class Batsman(TypedDict):
    runs : int
    balls : int
    sixes : int
    fours : int
    
    sr : float 
    bpb : float
    boundary_percent : float
    summart : str
```

```python
def calculate_sr(state : Batsman):
    strike_rate = (state['runs']/state['balls'])*100
    
    return {'sr' : strike_rate}
```

```python
def calculate_bpb(state : Batsman):
    bpb = state['balls']/(state['fours']+state['sixes'])
   
    return {'bpb' : bpb}
```

```python
def calculate_boundarypercent(state : Batsman):
    boundary_percent = (((state['fours']*4) + (state['sixes']*6))/state['runs'])*100
    
    return {'boundary_percent' : boundary_percent}
```

```python
def summary(state : Batsman) -> Batsman:
    summary = f"""
    Strike Rate - {state['sr']} \n
    Balls Per Boundary - {state['bpb']} \n
    Boundary Percent - {state['boundary_percent']}"""
    
    return {'summary' : summary}
```

```python
graph = StateGraph(Batsman)
graph.add_node('calculate_sr', calculate_sr)
graph.add_node('calculate_bpb', calculate_bpb)
graph.add_node('calculate_boundarypercent', calculate_boundarypercent)
graph.add_node('summary', summary)

graph.add_edge(START, 'calculate_sr')
graph.add_edge(START, 'calculate_bpb')
graph.add_edge(START, 'calculate_boundarypercent')

graph.add_edge('calculate_sr', 'summary')
graph.add_edge('calculate_bpb', 'summary')
graph.add_edge('calculate_boundarypercent', 'summary')

graph.add_edge('summary', END)

workflow = graph.compile()
```

```python
initial_state = {'runs' : 136, 'balls' : 64, 'sixes' : 8, 'fours' : 12 }
final_state = workflow.invoke(initial_state)

print(final_state)
```
