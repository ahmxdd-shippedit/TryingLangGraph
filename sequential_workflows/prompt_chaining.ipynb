```python
from langgraph.graph import StateGraph, START, END
from langchain_openai import ChatOpenAI
from typing import TypedDict
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
model = ChatOpenAI()
```

```python
class blog(TypedDict):
    title : str
    outline : str
    content : str
```

```python
def create_outline(state: blog) -> blog:
    title = state['title']
    prompt = f'Generate an outline for a blog on the title - {title}'
    
    outline = model.invoke(prompt).content
    state['outline'] = outline
    
    return state
```

```python
def create_blog(state:blog) -> blog:
    outline = state['outline']
    prompt = f'Generate a blog for this outline - {outline}'
    
    content = model.invoke(prompt).content
    state['content'] = content
    
    return state
```

```python
graph = StateGraph(blog)

graph.add_node('create_outline', create_outline)
graph.add_node('create_blog', create_blog)

graph.add_edge(START, 'create_outline')
graph.add_edge('create_outline', 'create_blog')
graph.add_edge('create_blog', END)

workflow = graph.compile()
```

```python
initial_state = {'title' : 'Why Drake actually won rap beef with Kendrick ?'}
final_state = workflow.invoke(initial_state)

print(final_state)
```
